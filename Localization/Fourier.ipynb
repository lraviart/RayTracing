```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.arange(0, 1, 1e-3)
x = np.ones_like(t)
X = np.fft.fft(x)
frequencies = np.fft.fftfreq(len(t), d=1e-3)

X = np.fft.fftshift(X)
frequencies = np.fft.fftshift(frequencies)

plt.figure(figsize=(8, 4))
plt.plot(frequencies, np.abs(X))
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("Fourier Transform of a Constant Signal")
plt.grid(True)
plt.show()
```
